```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from torchvision import transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader
```

```python
train_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.Grayscale(num_output_channels=1),
    transforms.ToTensor(),
])

eval_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.Grayscale(num_output_channels=1),
    transforms.ToTensor(),
])

train_ds = ImageFolder("./img/dataset/train", transform=train_tf)
val_ds = ImageFolder("./img/dataset/val", transform=eval_tf)
test_ds = ImageFolder("./img/dataset/test", transform=eval_tf)
```

```python
BATCH_SIZE = 4

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)

images, labels = next(iter(train_loader))
```

```python
import torch
import torch.nn as nn

device = "cuda" if torch.cuda.is_available() else "cpu"
num_classes = len(train_ds.classes)

class SimpleCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),   # 112
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),  # 56
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),  # 28
        )
        self.classifier = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(64, num_classes),
        )
    def forward(self, x):
        return self.classifier(self.features(x))

model = SimpleCNN(num_classes).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
```

```python
def run_epoch(loader, train=True):
    model.train(train)
    total_loss, correct, n = 0, 0, 0
    with torch.set_grad_enabled(train):
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            out = model(x)
            loss = criterion(out, y)
            if train:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
            total_loss += loss.item() * x.size(0)
            correct += (out.argmax(1) == y).sum().item()
            n += x.size(0)
    return total_loss / n, correct / n

history = {"tr_loss": [], "va_loss": [], "tr_acc": [], "va_acc": []}

EPOCHS = 10
for epoch in range(EPOCHS):
    tr_loss, tr_acc = run_epoch(train_loader, True)
    va_loss, va_acc = run_epoch(val_loader, False)
    for k, v in zip(history, [tr_loss, va_loss, tr_acc, va_acc]):
        history[k].append(v)
    print(f"Epoch {epoch+1}: train {tr_loss:.5f}/{tr_acc:.2%} | val {va_loss:.3f}/{va_acc:.2%}")
```

```python
import matplotlib.pyplot as plt

epochs = range(1, len(history["tr_loss"]) + 1)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

ax[0].plot(epochs, history["tr_loss"], label="train")
ax[0].plot(epochs, history["va_loss"], label="val")
ax[0].set_title("Loss")
ax[0].set_xlabel("epoch")
ax[0].legend()
ax[0].grid(alpha=0.3)

ax[1].plot(epochs, history["tr_acc"], label="train")
ax[1].plot(epochs, history["va_acc"], label="val")
ax[1].axhline(1/3, color="gray", linestyle="--", label="hasard (3 classes)")
ax[1].set_title("Accuracy")
ax[1].set_xlabel("epoch")
ax[1].legend()
ax[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay

classes = train_ds.classes

# Répartition des classes par split
for name, ds in [("train", train_ds), ("val", val_ds), ("test", test_ds)]:
    c = Counter(ds.targets)
    print(name, {classes[k]: v for k, v in sorted(c.items())})

# Prédictions sur le test
def get_preds(loader):
    model.eval()
    preds, trues, probs = [], [], []
    with torch.no_grad():
        for x, y in loader:
            out = model(x.to(device))
            probs += torch.softmax(out, 1).cpu().tolist()
            preds += out.argmax(1).cpu().tolist()
            trues += y.tolist()
    return np.array(trues), np.array(preds), np.array(probs)

y_true, y_pred, y_prob = get_preds(test_loader)

# Rapport : precision / recall / f1 par classe
print(classification_report(y_true, y_pred, target_names=classes, digits=3))

# Matrice de confusion
ConfusionMatrixDisplay(confusion_matrix(y_true, y_pred), display_labels=classes).plot(cmap="Blues")
plt.title("Matrice de confusion (test)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def show_predictions(ds, y_true, y_pred, y_prob, indices, cols=6):
    rows = int(np.ceil(len(indices) / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 2.5, rows * 3))
    axes = np.array(axes).reshape(-1)
    for ax in axes:
        ax.axis("off")
    for ax, i in zip(axes, indices):
        img, _ = ds[i]
        ok = y_true[i] == y_pred[i]
        ax.imshow(img.squeeze(), cmap="gray")
        ax.set_title(
            f"vrai: {classes[y_true[i]]}\n"
            f"prédit: {classes[y_pred[i]]} ({y_prob[i][y_pred[i]]:.0%})",
            fontsize=9, color="green" if ok else "red",
        )
    plt.tight_layout()
    plt.show()

# 12 images au hasard du test set
rng = np.random.default_rng(0)
show_predictions(test_ds, y_true, y_pred, y_prob,
                 rng.choice(len(test_ds), size=min(12, len(test_ds)), replace=False))
```
